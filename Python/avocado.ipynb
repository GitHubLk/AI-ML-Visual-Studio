```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np



df = pd.read_csv('avocado.csv', index_col=0)
print("Dataset read successfully into DataFrame 'df'.")
    
    #Step2: Reading the Dataset
    # Check total missing values before dropping
missing_count_before = df.isnull().sum().sum()
print(f"\nTotal missing values found: {missing_count_before}")
    
    # Drop rows with any missing values - The one-line hint

df_cleaned = df.dropna()
    
    # Update the reference to the cleaned DataFrame
df = df_cleaned
    
    # Verify the results
missing_count_after = df.isnull().sum().sum()
print(f"Total missing values after dropna(): {missing_count_after}")
print(f"DataFrame shape after cleaning: {df.shape}")
print("\nFirst 5 rows of the processed DataFrame:")
print(df.head())



#Step3: Extract Features
excluded_cols = ['region', 'Date']

# Extract all features by dropping the excluded columns
# The resulting DataFrame contains all numerical and categorical features except 'region' and 'Date'.
feature_df = df.drop(columns=excluded_cols)

print("Feature Extraction Complete!")
print(f"Original DataFrame columns: {df.columns.tolist()}")
print(f"Excluded Columns: {excluded_cols}")
print(f"Features DataFrame columns: {feature_df.columns.tolist()}")
print("\nFirst 5 rows of the Features DataFrame:")
print(feature_df.head())



#Step4: 
target_col = 'AveragePrice'

# Identify numerical and categorical columns for pre-processing
# 'type' is the only remaining categorical feature.
numerical_features = feature_df.select_dtypes(include=np.number).columns.tolist()
categorical_features = feature_df.select_dtypes(include='object').columns.tolist()

# Remove the target from the features lists
numerical_features.remove(target_col)
# The 'Date' and 'region' columns were already removed.

print("--- Feature Identification ---")
print(f"Numerical Features to Scale: {numerical_features}")
print(f"Categorical Features to Encode: {categorical_features}")

# 2. Build the Pre-processing Pipeline using ColumnTransformer
# A ColumnTransformer is used to apply different transformations to different columns.
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)
    ],
    remainder='passthrough' # Keep any columns not explicitly transformed (shouldn't be any here)
)

# 3. Apply the Pre-processing to the Features
# Separate the features (X) and the target (y)
X = feature_df.drop(columns=[target_col])
y = feature_df[target_col]

# Apply the transformations to the features
X_processed = preprocessor.fit_transform(X)

# 4. Convert the result back to a DataFrame for easier inspection
# Get the names of the columns after One-Hot Encoding
encoded_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
all_feature_names = numerical_features + list(encoded_feature_names)

X_processed_df = pd.DataFrame(
    X_processed,
    columns=all_feature_names,
    index=X.index
)

print("\n--- Pre-processing Summary ---")
print(f"Features (X) shape before processing: {X.shape}")
print(f"Features (X) shape after processing: {X_processed_df.shape}")
print("\nFirst 5 rows of the Processed Features DataFrame (X_processed_df):")
print(X_processed_df.head())



#5 Split your data as follows:

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X_processed_df,
    y,
    test_size=0.1,  # 10% for the final test set
    random_state=42 # Ensures reproducibility
)

print("--- Step 1: Test Set Separation ---")
print(f"Test Set Shape (10%): {X_test.shape}")
print(f"Train/Validation Combined Shape (90%): {X_train_val.shape}")

# 2. Second Split: Separate Training (8/9 of remaining) and Validation (1/9 of remaining)
# The remaining X_train_val (90% of original) needs to be split 80% (Train) / 10% (Validation).
# To get 10% of the *original* data for validation, we must split the *remaining 90%* by 1/9.
# 0.10 / 0.90 = 0.1111... (approx 11.11%)
validation_split_ratio = 0.10 / 0.90

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=validation_split_ratio, # 1/9 of the remaining data
    random_state=42
)

print("\n--- Step 2: Train/Validation Set Separation ---")
print(f"Training Set Shape (80%): [X_train.shape]")
print(f"Validation Set Shape (10%): {X_val.shape}")




#Step6:

k_value = 5
knn_model = KNeighborsRegressor(n_neighbors=k_value)

print(f"Initializing KNN Regressor with k = {k_value}...")

# 2. Train the Model (Fit)
# We train the model using the training data.
knn_model.fit(X_train, y_train)

print(" Model training complete.")

# 3. Evaluate Model Performance

# A. Make Predictions on the Validation Set
y_val_pred = knn_model.predict(X_val)

# B. Calculate Metrics for Validation Set
val_mse = mean_squared_error(y_val, y_val_pred)
val_rmse = np.sqrt(val_mse)
val_r2 = r2_score(y_val, y_val_pred)

print("\n--- Validation Set Performance (k=5) ---")
print(f"Mean Squared Error (MSE): {val_mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {val_rmse:.4f}")
print(f"R-squared (R²): {val_r2:.4f}")

# C. Optional: Evaluate on the Training Set (for baseline comparison)
y_train_pred = knn_model.predict(X_train)
train_r2 = r2_score(y_train, y_train_pred)

print(f"Training Set R-squared (R²): {train_r2:.4f}")

k_values = range(1, 21)
validation_mse = []

for k in k_values:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_val_pred = knn.predict(X_val)
    validation_mse.append(mean_squared_error(y_val, y_val_pred))

best_k = k_values[np.argmin(validation_mse)]
print(f"Best k on validation set: {best_k}")

# 2. Train final KNN on training + validation
X_train_full = np.vstack((X_train, X_val))
y_train_full = np.hstack((y_train, y_val))

knn_final = KNeighborsRegressor(n_neighbors=best_k)
knn_final.fit(X_train_full, y_train_full)

# 3. Predict on test set and compute R-squared
y_test_pred = knn_final.predict(X_test)
r2 = r2_score(y_test, y_test_pred)
print(f"Final KNN Regressor R² on test set: {r2:.4f}")


```
